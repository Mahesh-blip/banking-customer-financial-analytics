```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

#Missing Values
customers.isnull().sum()

#Duplicate Rows
customers.duplicated().sum()

#Unique Values
customers.nunique()

#Gender Values
customers["gender"].value_counts()

#Standardize Gender
customers["gender"] = customers["gender"].str.title()

#Check Gender Values again
customers["gender"].value_counts()

#Fill Missing City with Unknown
customers["city"] = customers["city"].fillna("Unknown")

#Remove Duplicates
customers = customers.drop_duplicates()

#Convert Date
customers["customer_since"] = pd.to_datetime(customers["customer_since"])

#Age Distribution
plt.figure(figsize=(8,5))

plt.hist(customers["age"], bins=15)

plt.title("Age Distribution")

plt.show()
```

```python
#Gender Distribution
customers["gender"].value_counts().plot(kind="bar")

plt.show()
```

```python
#City-wise Customers*
customers["city"].value_counts().head(10).plot(kind="bar")
```

```python
#Occupation-wise Customers*
customers["occupation"].value_counts().plot(kind="bar")
```

```python
#Phase 5 — Feature Engineering*
#Create New Column called Age Groups*
customers["Age_Group"] = cut(
    customers["age"],
    bins=[18,25,35,45,60,100],
    labels=["18-25","26-35","36-45","46-60","60+"]
)
```

```python
#Add New column Customer Tenure*
today = Timestamp.today()

customers["Tenure"] = (
    today - customers["customer_since"]
).dt.days/365
```

```python
#Phase 6 — KPI Analysis*

#Total Customers*
customers["customer_id"].nunique()

#Average Income*
customers["annual_income"].mean()

#Highest Income*
customers["annual_income"].max()

#Average Balance*
accounts["balance"].mean()

#Total Loan Amount*
loans["loan_amount"].sum()

#Average Credit Score*
loans["credit_score"].mean()
```

```python
print("========== BANK KPI DASHBOARD ==========")

print("Total Customers :", customers["customer_id"].nunique())

print("Total Accounts :", accounts["account_id"].nunique())

print("Total Transactions :", len(transactions))

print("Total Balance : ₹{:,.2f}".format(accounts["balance"].sum()))

print("Loan Portfolio : ₹{:,.2f}".format(loans["loan_amount"].sum()))

print("Average Balance : ₹{:,.2f}".format(accounts["balance"].mean()))

print("Average Credit Score :", round(loans["credit_score"].mean()))

print("Loan Default Rate : {:.2f}%".format(
    (loans["default_status"]=="Yes").mean()*100
))
```

```python
#Dashboard Visualizations
#Chart 1 – Customer Age Groups*

plt.figure(figsize=(8,5))

customers["Age_Group"].value_counts().sort_index().plot(kind="bar")

plt.title("Customer Age Groups")
plt.xlabel("Age Group")
plt.ylabel("Customers")

plt.show()

"""
Business Insight

Most customers belong to the working-age population, making them ideal candidates for loans and investment products.
"""
```

```python
#Chart 2 – Income Category*

customers["annual_income"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.ylabel("")
plt.title("Income Distribution")

plt.show()
```

```python
#Chart 3 – Account Balance Distribution*

plt.figure(figsize=(8,5))

sns.histplot(accounts["balance"], bins=30)

plt.title("Account Balance Distribution")

plt.show()
```

```python
#Chart 4 – Monthly Transaction Trend*
transactions["txn_date"] = to_datetime(transactions["txn_date"])
transactions["month"] = transactions["txn_date"].dt.month_name()

monthly = transactions.groupby("month")["amount"].sum()

monthly.plot(
    figsize=(12,5),
    marker="o"
)

plt.title("Monthly Transaction Trend")

plt.show()
```

```python


def risk(score):
    if score >= 750:
        return "Low Risk"
    elif score >= 650:
        return "Medium Risk"
    else:
        return "High Risk"

loans["risk_category"] = loans["credit_score"].apply(risk)
```

```python
#Chart 7 – Risk Category*

loans["risk_category"].value_counts().plot(
    kind="bar"
)

plt.title("Risk Categories")

plt.show()
```

```python

accounts["customer_segment"] = cut(
    accounts["balance"],
    bins=[0, 50000, 200000, 500000, float("inf")],
    labels=["Regular", "Silver", "Gold", "Premium"]
)
```

```python
#Chart 8 – Customer Segments
accounts["customer_segment"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.ylabel("")
plt.title("Customer Segments")

plt.show()
```

```python
#Chart 9 – Branch Performance*
branch = accounts.groupby("branch_id")["balance"]\
.sum()\
.sort_values(ascending=False)\
.head(10)

branch.plot(kind="bar")

plt.title("Top 10 Branches")

plt.show()
```

```python
customer_accounts = pd.merge(
    customers,
    accounts,
    on="customer_id",
    how="inner"
)
```

```python
#Chart 10 – Occupation-wise Balance*
customer_accounts.groupby("occupation")["balance"]\
.mean()\
.sort_values()\
.plot(kind="barh")

plt.title("Average Balance by Occupation")

plt.show()
```

```python
occupation_balance = (
    customer_accounts
    .groupby("occupation")["balance"]
    .mean()
    .sort_values()
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

occupation_balance.plot(kind="barh")

plt.title("Average Balance by Occupation")
plt.xlabel("Average Balance")
plt.ylabel("Occupation")

plt.show()
```
